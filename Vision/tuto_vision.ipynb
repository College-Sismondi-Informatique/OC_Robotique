## (Rappel) sur les listes

!!! info Les listes en Python

En Python, une liste est une structure de données qui contient une collection d'objets. Il s'agit d'un nouveau type par rapport aux entiers, float, booléens et chaînes de caractères.

Les listes sont définies avec des crochets `[ ]`et les éléments sont séparés par des virgules:

``` python
# Liste de nombres
nombres = [1, 2, 3, 4, 5]

# Liste de chaînes de caractères
fruits = ["pomme", "banane", "orange"]

# Liste mixte (hétérogène)
mixte = [1, "hello", 3.14, True]

# Liste vide
vide = []
```
!!!

!!! info Accès aux éléments

On accède aux éléments de la liste avec des crochets `[ ]` qui contiennent l'index de l'élément dans la liste.


``` python
print(fruits[2])

mixte[3] = 12
```
!!!

!!! tip Début de l'index

**Attention :** l'index commence à 0 : `fruits[0]` permet d'acceder au 1er élément et `fruits[2]` au 3ème.

!!! info Parcourir une liste

En python, on peut parcourir avec une boucle qui itère sur son index : 


Mais on peut également parcourir directement les éléments de la liste : 

``` python
for element in mixte: 
    print(element)
```

!!! info Flou et masque de netteté
#### Flou
Le flou sur une image est obtenu en lissant les contours, en faisant une moyenne des pixels environnants. Sur openCV, on peut utiliser la fonction `medianBlur` pour flouter une image.

⚠️ La fonction `medianBlur` prend 2 paramètres, l'image et la taille du filtre. Ce second paramètre doit être un nombre **impair supérieur à 1**.


#### Masque de netteté
On peut ensuite soustraire l'image floutée à l'image originale pour lui faire un masque de netteté.

Pour finir, on va prendre 2 photos du même objets et essayer de faire correspondre des points d'intérêts entre les 2, à l'aide de l'algorithme [scale-invariant feature transform (SIFT)](https://fr.wikipedia.org/wiki/Scale-invariant_feature_transform).

!!! note Exercice 7

2) Utiliser ensuite le programme suivant pour détecter et faire correspondre les points d'intérêt.

```python
# Initiate SIFT detector
sift = cv2.SIFT_create()
 
# find the keypoints and descriptors with SIFT
kp1, des1 = sift.detectAndCompute(img1,None)
kp2, des2 = sift.detectAndCompute(img2,None)
 
# BFMatcher with default params
bf = cv2.BFMatcher()
matches = bf.knnMatch(des1,des2,k=2)
 
# Apply ratio test
good = []
for m,n in matches:
    if m.distance < 0.75*n.distance:
        good.append([m])
        
# cv.drawMatchesKnn expects list of lists as matches.
img3 = cv2.drawMatchesKnn(img1,kp1,img2,kp2,good,None,flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)
```
