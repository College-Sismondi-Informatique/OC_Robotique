```python
import cv2


image = cv2.imread("arucos_robot.png")

cv2.imshow('Detected Markers', image)
```

```python
aruco_dict = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_250)
parameters = cv2.aruco.DetectorParameters()

detector = cv2.aruco.ArucoDetector(aruco_dict, parameters)
```

```python
# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow('Gray image', gray)
```

```python

# Detect the markers
corners, ids, rejected = detector.detectMarkers(gray)

image_with_markers = image.copy()
if ids is not None:
    print(corners)
    cv2.aruco.drawDetectedMarkers(image_with_markers, corners, ids)
    
cv2.imshow('Detected Markers', image_with_markers)
```

```python
import numpy as np

n = 0

def orientation(A,B):
        dx = B[0] - A[0]
        dy = B[1] - A[1]

        # Calculer l'angle en radians
        angle_rad = np.arctan2(dy, dx)

        # Convertir en degrés
        angle_deg = np.degrees(angle_rad)
        return angle_deg

print(orientation(corners[n][0][1],corners[n][0][0]))
```
