```python
# Votre code ici
import cv2

image = cv2.imread("arucos.png")

cv2.imshow('Original photo', image)

# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow('Gray image', gray)
```

!!! note Exercice 2
Cherchez en ligne les fonctions d'openCV permettant de détecter des marqueurs ARUCO, puis remplir le programme suivant qui doit : 

* Créer le dictionnaire de marqueurs et le détecteur,
* Détecter les marqueurs,
* Dessiner les marqueurs sur l'image et l'afficher.


!!!

```python
# Créer le dictionnaire de marqueurs et le détecteur
aruco_dict = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_250)

parameters = cv2.aruco.DetectorParameters()
detector = cv2.aruco.ArucoDetector(aruco_dict, parameters)

# Détecter les marqueurs
corners, ids, rejected = detector.detectMarkers(gray)


# Dessiner les marqueurs sur l'image et l'afficher
image_with_markers = image.copy() # pour éviter de dessiner sur l'image
if ids is not None:
    print(corners)
    cv2.aruco.drawDetectedMarkers(image_with_markers, corners, ids)
    
cv2.imshow('Detected Markers', image_with_markers)
```

```python
import numpy as np

def position(corner):
    return np.mean(corner, axis=0) # moyenne verticalement

corner = corners[0][0]

print(position(corner))
```

!!! note Exercice 4b
3) Réfléchir à l'aide d'un papier et d'un crayon à comment calculer l'orientation d'un marqueur à partir de ses coins.
4) Créer une fonction `orientation(corner)` qui calcule et renvoie l'orientation en degrés d'un marqueur.
!!!

```python
import math

def orientation(corner):
    dx = corner[0][0] - corner[1][0]
    dy = corner[0][1] - corner[1][1]

    # Calculer l'angle en radians
    angle_rad = math.atan2(dy, dx)

    # Convertir en degrés
    angle_deg = math.degrees(angle_rad)
    return angle_deg

corner = corners[0][0]

print(orientation(corner))
```
